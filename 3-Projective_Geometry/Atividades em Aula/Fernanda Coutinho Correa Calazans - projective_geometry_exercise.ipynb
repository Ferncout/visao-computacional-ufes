###2- Determine whether or not the point "p" belongs to the line "r":

a) r = (3, 5, 2), p = (1, -1, 1)

b) r = (3, 5, 2), p = (1, 0, 1)

c) r = (3, 5, 2), p = (1, 1, 1)

d) r = (2, 1, -6), p = (-2, 0, 1)

e) r = (2, 1, -6), p = (-2, -4, 1)

f) r = (-4, 7, 0), p = (5, -3, 1)

g) r = (-4, 7, 0), p = (14, 28, 3)

h) r = (0, 0, 1), p = (0, 0, 1)

i) r = (0, 0, 1), p = (4, 0, 0)

Follow the code example below:

```
#line
r = np.array([3, 3, 1])
#point
p = np.array([1, 1, 2])
#Calculate the inner product and check if it equals zero.
d = np.dot(r,p)
print(d)
if d == 0:
    print("The point belongs to the line.")
else:
    print("The point doesn't belong to the line.")
```

```
# definindo uma função
def belong(r, p, q):
  d = np.dot(r, p)
  if d == 0:
    return print(f"{q}) The point {p} belongs to the line")
  else:
    return print(f"{q}) The point {p} belongs to the line")

# a)
r = np.array([3, 5, 2])
p = np.array([1, -1, 1])
belong(r, p, 'a')

# b)
r = np.array([3, 5, 2])
p = np.array([1, 0, 1])
belong(r, p, 'b')

# c)
r = np.array([3, 5, 2])
p = np.array([1, 1, 1])
belong(r, p, 'c')

# d)
r = np.array([2, 1, -6])
p = np.array([-2, 0, 1])
belong(r, p, 'd')

# e)
r = np.array([2, 1, -6])
p = np.array([-2, 4, 1])
belong(r, p, 'e')

# f)
r = np.array([-4, 7, 0])
p = np.array([5, -3, 1])
belong(r, p, 'f')

# g)
r = np.array([-4, 7, 0])
p = np.array([14, 28, 3])
belong(r, p, 'g')

# h)
r = np.array([0, 0, 1])
p = np.array([0, 0, 1])
belong(r, p, 'h')

# i)
r = np.array([0, 0, 1])
p = np.array([4, 0, 0])
belong(r, p, 'i')
```

###3- Determine the line that passes through the points:

a) (2, 3, 1) e (5, 6, 4)

b) (-3, 1, -2) e (3, 2, -1)

c) (8, -3, -1) e (-7, 9, 0)

Follow the code example below:


```
#points
p1 = np.array ([10,-3, 1])
p2 = np.array ([-3, 5, 1])

#calculate the line

r = np.cross(p1,p2)
print("The line is ", r)
```

```
# definindo uma função
def line(p1,p2,q):
  # r = p1@p2 (pq não funciona?)
  r = np.cross(p1,p2)
  return print(f"{q}) The line is {r}")

# a)
p1 = np.array([2, 3, 1])
p2 = np.array([5, 6, 4])
line(p1,p2, 'a')

# b)
p1 = np.array([-3, 1, -2])
p2 = np.array([-3, 2, -1])
line(p1,p2, 'b')

# c)
p1 = np.array([8, -3, -1])
p2 = np.array([-7, 9, 0])
line(p1,p2, 'c')
```

###4- Determine the point where the following pairs of lines intersect:

a)  (1, 0, 0) e (0, 1, 0)

b)  (3, 5, 1) e (4, 6, 2)

c)  (3, 5, 1) e (0, 0, 1)

d)  (3, 5, 1) e (3, 5, 2)

Follow the code example below:

```
# lines
r1 = np.array([4, 7, 3])
r2 = np.array([-3, 1, -5])

# intersection point
p = np.cross(r1,r2)
p = p/p[2]
print("The intersection point is ", p)
```

```
# definindo a função
def point(r1, r2, q):
  p = np.cross(r1, r2)
  if p[-1] == 0:
    return print(f"{q}) The intersection point is {p}")
  else:
    p = p/p[-1]
    return print(f"{q}) The intersection point is {p}")

# a)
r1 = np.array([1,0,0])
r2 = np.array([0,1,0])
point (r1,r2, 'a')

# b)
r1 = np.array([3,5,1])
r2 = np.array([4,6,2])
point (r1,r2, 'b')

# c)
r1 = np.array([3,5,1])
r2 = np.array([0,0,1])
point (r1,r2, 'c')

# d)
r1 = np.array([3,5,1])
r2 = np.array([3,5,2])
point (r1,r2, 'd')
```
